# 05 - Machine Learning Anomaly Detection
Training Isolation Forest for unsupervised configuration pattern detection and Random Forest risk classifier.

```python
from sklearn.ensemble import IsolationForest, RandomForestClassifier
import pandas as pd

df = pd.read_csv('../data/processed/ml_features.csv')
X = df[['telnet_num', 'ssh_num', 'logging_num', 'guest_iso_num', 'dhcp_snoop_num', 'security_weakening_score', 'segmentation_risk_score', 'has_authorized_ticket', 'num_changed_fields']]

iso = IsolationForest(contamination=0.05, random_state=42)
iso.fit(X)
print('Isolation forest trained on', len(X), 'samples.')
```
