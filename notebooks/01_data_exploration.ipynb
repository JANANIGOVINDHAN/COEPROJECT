# 01 - Data Exploration
Exploratory data analysis of synthetic hospital network device configurations, sites, events, and change tickets.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

devices = pd.read_csv('../data/raw/devices_raw.csv')
configs = pd.read_csv('../data/raw/configurations_raw.csv')
tickets = pd.read_csv('../data/raw/change_tickets_raw.csv')

print('Raw Devices Count:', len(devices))
print('Raw Configurations Count:', len(configs))
print('Raw Change Tickets Count:', len(tickets))
devices.head()
```

```python
configs['device_type'].value_counts().plot(kind='bar', title='Configuration Snapshots by Device Type')
plt.show()
```
